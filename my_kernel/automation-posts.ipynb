```python
!pip install -U discord.py

##### 🧭 Instalo Playwright y sus navegadores
!pip install --no-cache-dir playwright==1.45.0
!playwright install

# =========================
# 📦 Imports estándar Python
# =========================

import os                    # 🗂️ Manejo del sistema de archivos (paths, variables de entorno, etc.)
import json                  # 📄 Lectura y escritura de datos en formato JSON
import time                  # ⏱️ Manejo de tiempos, delays y timestamps
import random                # 🎲 Generación de valores aleatorios
import re                    # 🔍 Expresiones regulares para búsqueda y limpieza de texto
import asyncio               # 🔁 Programación asíncrona con async / await
import tempfile              # 🧪 Creación de archivos temporales
import urllib.parse          # 🌐 Codificación y decodificación de URLs
from datetime import datetime # 📅 Manejo de fechas y horas
from math import gcd         # ➗ Cálculo del máximo común divisor (útil para proporciones / imágenes)
import imghdr                 # 🖼️ Detección del tipo/formato real de imágenes a partir de su contenido

# =========================
# 🌐 Requests & HTTP
# =========================

import requests              # 🌍 Peticiones HTTP síncronas (APIs, descargas, etc.)
import aiohttp               # ⚡ Cliente HTTP asíncrono (más rápido para múltiples requests)

# =========================
# 🖼️ Imágenes & Display (Kaggle / Jupyter)
# =========================

from PIL import Image        # 🖼️ Manipulación de imágenes (abrir, guardar, redimensionar)
from io import BytesIO       # 📦 Manejo de imágenes/datos binarios en memoria
from IPython.display import display  # 👀 Mostrar imágenes u objetos visuales en notebooks

# =========================
# 🔐 OAuth & Autenticación
# =========================

from requests_oauthlib import OAuth1      # 🔑 Autenticación OAuth 1.0 (Twitter/X clásico, etc.)

# =========================
# 📊 Google APIs
# =========================

import gspread                              # 📈 Acceso y manipulación de Google Sheets
from google.oauth2 import service_account  # 🔐 Autenticación con Service Account
from googleapiclient.discovery import build # 🧩 Construcción de clientes para Google APIs

# =========================
# 🐦 Twitter / X
# =========================

import tweepy                # 🐦 SDK oficial para la API de Twitter/X

# =========================
# 💬 Discord
# =========================

import discord               # 💬 Bot de Discord (mensajes, embeds, eventos)

# =========================
# 🕸️ Playwright – Automatización Web
# =========================

from playwright.sync_api import sync_playwright
# 🧭 Playwright síncrono:
# - Navegación web automatizada
# - Ideal para scripts simples o debugging rápido
# - Compatible con Chromium, Firefox y WebKit

from playwright.async_api import async_playwright, TimeoutError
# ⚡ Playwright asíncrono:
# - Ideal para scraping avanzado y automatización concurrente
# - Mejor rendimiento con múltiples páginas / tareas
# - TimeoutError para controlar fallos de carga

# =========================
# 🔁 Async helpers (Notebooks)
# =========================

import nest_asyncio
# 🔄 Permite ejecutar asyncio dentro de entornos que ya tienen un event loop
# (Kaggle, Jupyter, Colab)

nest_asyncio.apply()  # ✅ Muy importante para evitar errores tipo "event loop is running"

# =========================
# 🔐 Kaggle Secrets
# =========================

from kaggle_secrets import UserSecretsClient
# 🔑 Acceso seguro a API keys y tokens guardados en Kaggle
# (Twitter, Discord, OpenAI, etc.)


user_secrets = UserSecretsClient()

# Credenciales de API de Google Sheets
credentials_info = {
    "type": "service_account",
    "project_id": user_secrets.get_secret("GOOGLE_SHEETS_PROJECT_ID"),
    "private_key_id": user_secrets.get_secret("GOOGLE_SHEETS_PRIVATE_KEY_ID"),
    "private_key": "-----BEGIN PRIVATE KEY-----\nMIIEuwIBADANBgkqhkiG9w0BAQEFAASCBKUwggShAgEAAoIBAQDI5TCyAQsRnBGM\nYt1q5gb5ANPkqRzEOcKovNT+j5bDvbkjb91KivbPLlqYNPKi+H+Va3si9+3ImpOa\nOKlGwJYdSWPMqA8VWp7/WuiC0exKZnC3WL7Bos9ODMmwAQWkllRu3ZST8SASKGwW\n3D+XLZog0UGcGFWTn4rEjjeD6sTCwFeYIAo2+dpae44i283tw+s2Kf40Mdc3qUXO\n3E6GEvkjfAgChXYy1LmVo7BRKeAzKq75mW3Zme5EgwjeYyCRif5+NPCrGypRaVDb\nZVGn9SsuVInI6cxmF687undNj8jgaRiX5dGK8bQv3SlVCV+aGqbLV/WNT/+uGNXw\n8NajfTp5AgMBAAECggEABEdTq8zczfJAHO3GR7aFt5QCZKG16+zLgKwXO2M2llgS\nIKc1+Yu/YS78bWFeTrqCaXMlh0YsGCxI9b1HHKtxdFxPE0l72xbH9rS6fukb563q\nG2Y5AiyBTrPJcAr7h5N55vZnlFA2axtANGtFMZyeO7HykZGim/SR9ZOjqswgm+xd\nX9e8ucSQlstFcbXYShbLg/NTFJ9/qkLL3Nru5glLC4mhpnbrofsS0DFuemWdCOEQ\nTXRQMZTj1kEQKoLbVWLaldNLI4pa+arY4K/+VJ+CwBS6INGHe3Jt7cTbc6BKwRRt\nK2cZrLYIc3gCyO+OJc2tK/L/jlzTcsuZeUh4hjcZUQKBgQDnEA21gIvdVOYiovOT\nmdXHrWCiBVWpn0dc+kzsz6BaqRz0b1QN7hOwG/shRrGXzjAAIWioOJKFfCvNwIyZ\nhxlGIzhcgijJl65/H4O1MfTdleOaAnBvz/EJ6j76KhH0LDNHTiC3Fil/G4ebflkk\nRb9PXDt9vmJKqLJbInJC8Bz+6QKBgQDek6bXTGt2UaPGF20J3dxf996nF5NZrids\n5RLenJ7pnWSZBnW7ZrsZRqBARJIUWomX9/e8QZpdcDVYdqwaxRutZvAk5FxYyfah\n+u+xLhSncuR1+9UUW8g6fglQlrYe+jF6G9hsZ9Q2utKTqfs5DiDapbploWtFA0Bz\nFELOS8/FEQKBgDHZgculpwciOOBCHaC7cKa6nf12fyuwJFpH2e7BQIwurgOCPTUg\nJvk4oC5mrQvePPbaPlUEl3/a8SxbnTnGLsq18IPsxNXxNoFQkD64P2bK3qasXguR\nZiRIeDvKdcMjAhEhJue5DUzR6KjAUoUD0ylQ8bjIKTaIPKpyvQn5+Zd5An9aa6Fb\n1a/SYVgnIV0qpPbiBdBuESXLSPS7SFRiGz5zqX00rWsmtpVLr2VqtnCh1CYRlqsN\nz5jmXK9PQOap+WlFCIyGcD8T0vhe1ZKKaC372LD3PVdS7fZpzMn78z4JV5rCAI3r\nfEK5+SuS0Xs9sn+mU61lnC3321pTZee95w6BAoGBAMOVuM+79x57RLrv1JpL83+d\nX7GoljXABiswOTCxL1Gp7hN9OqiVNSheAR+nLAQ+ctKtM40gvQIA46N38HF+ptiB\ny17Rh1ZWmAmtW0qGXahMjJ7f2wsCGmqVz5i9xfsPjNkWPNfCVwVxGb696aHTMbJT\nmdFDueDrx+UhmMI1vFoF\n-----END PRIVATE KEY-----",
    "client_email": user_secrets.get_secret("GOOGLE_SHEETS_CLIENT_EMAIL"),
    "client_id": user_secrets.get_secret("GOOGLE_CLIENT_ID"),
    "auth_uri": "https://accounts.google.com/o/oauth2/auth",
    "token_uri": "https://oauth2.googleapis.com/token",
    "auth_provider_x509_cert_url": "https://www.googleapis.com/oauth2/v1/certs",
    "client_x509_cert_url": "https://www.googleapis.com/robot/v1/metadata/x509/python-api%40automatizaciones-449817.iam.gserviceaccount.com",
    "universe_domain": "googleapis.com"
}

# Credenciales de API de Twitter
TWITTER_API_KEY = user_secrets.get_secret("TWITTER_API_KEY")
TWITTER_API_SECRET_KEY = user_secrets.get_secret("TWITTER_API_SECRET_KEY")
TWITTER_ACCESS_TOKEN =user_secrets.get_secret("TWITTER_ACCESS_TOKEN")
TWITTER_ACCESS_TOKEN_SECRET = user_secrets.get_secret("TWITTER_ACCESS_TOKEN_SECRET")

# Autenticación OAuth1 para Twitter
auth = OAuth1(TWITTER_API_KEY, TWITTER_API_SECRET_KEY, TWITTER_ACCESS_TOKEN, TWITTER_ACCESS_TOKEN_SECRET)

# Parámetros de autenticación para Twitter
TWITTER_BEARER_TOKEN = user_secrets.get_secret("TWITTER_BEARER_TOKEN")

# V2 Twitter API Authentication
client = tweepy.Client(
    consumer_key=TWITTER_API_KEY,
    consumer_secret=TWITTER_API_SECRET_KEY,
    access_token=TWITTER_ACCESS_TOKEN,
    access_token_secret=TWITTER_ACCESS_TOKEN_SECRET
)

# API KEY de NEWS API
NEWS_API_KEY = user_secrets.get_secret("NEWS_API_KEY")

# API KEY de GNEWS API
GNEWS_API_KEY = user_secrets.get_secret("GNEWS_API_KEY")

# API KEY de OPENROUTER_API_KEY API
OPENROUTER_API_KEY = user_secrets.get_secret("OPENROUTER_API_KEY")

# API KEY de OPENAI API
OPENAI_API_KEY = user_secrets.get_secret("OPENAI_API_KEY")

# TOKEN del BOT DISCORD
DISCORD_TOKEN_BOT = user_secrets.get_secret("DISCORD_TOKEN_BOT")

# Habilito intents adicionales
intents = discord.Intents.default()
intents.messages = True  # Necesario para leer los mensajes
intents.guilds = True  # Necesario para acceder a la información del servidor

# Creo el cliente de Discord con los intents
discord_client = discord.Client(intents=intents)

# URL del Webhook de Make
webhook_url = "https://hook.eu2.make.com/c6x6to2ja1vty98h392v59rtuityhyb8"

# ---------------------------Google Sheets----------------------------------
# Autenticación con Google Sheets y obtengo el Worksheet
def get_worksheet():
    try:
        scopes = ['https://www.googleapis.com/auth/spreadsheets']
        credentials = service_account.Credentials.from_service_account_info(credentials_info, scopes=scopes)
        gc = gspread.authorize(credentials)
        sheet_id = "10ki86a5Z-UUwmtAD5QWfTV9XqzgEYdTPWhJ1cX_hdTE"
        spreadsheet = gc.open_by_key(sheet_id)
        worksheet = spreadsheet.sheet1
        return worksheet
    except Exception as e:
        print(f"Error al autenticar o acceder a Google Sheets: {e}")
        return None

# Leo las noticias publicadas
def leer_noticias_publicadas():
    noticias_publicadas = set()
    try:
        worksheet = get_worksheet()
        if not worksheet:
            print("No se pudo obtener el worksheet.")
            return noticias_publicadas
        
        filas = worksheet.get_all_values()
        for fila in filas:
            if fila:
                noticias_publicadas.add(fila[0])  # Las URLs están en la primera columna
    except Exception as e:
        print(f"Error al leer las noticias publicadas: {e}")
    return noticias_publicadas

# Guardo la noticia publicada
def guardar_noticia_publicada(url):
    try:
        worksheet = get_worksheet()
        if not worksheet:
            print("No se pudo obtener el worksheet para guardar la noticia.")
            return

        total_filas = len(worksheet.get_all_values())
        
        # Si hay más de 10 filas, borrar todas
        if total_filas > 10:
            worksheet.clear()
            print("🧹 Worksheet limpiado porque había más de 10 filas.")
        
        # Agregar la nueva URL como primera fila
        worksheet.insert_row([url], 1)
        print(f"✅ URL guardada exitosamente en la primera fila: {url}")
        
    except Exception as e:
        print(f"Error al guardar la noticia publicada: {e}")

# Obtengo noticias guardadas desde Google Sheets
def obtener_noticias_guardadas():
    try:
        worksheet = get_worksheet()
        if not worksheet:
            print("No se pudo obtener el worksheet.")
            return []

        filas = worksheet.get_all_values()

        # Extraer URLs válidas de la primera columna
        noticias_guardadas = [
            fila[0] for fila in filas
            if fila and fila[0].strip() != ""
        ]

        return noticias_guardadas

    except Exception as e:
        print(f"Error al obtener las noticias guardadas: {e}")
        return []

# ---------------------------Noticia----------------------------------
# Función principal para obtener noticia validada
def obtener_noticia(noticias_guardadas, max_noticias=10):
    noticias_obtenidas = []

    # Intento obtener hasta el max_noticias
    while len(noticias_obtenidas) < max_noticias:
        noticia = obtener_noticia_newsapi(noticias_guardadas)
        if not noticia:
            noticia = obtener_noticia_gnews(noticias_guardadas)
        if not noticia:
            print("❌ No se encontraron noticias nuevas.")
            break

        if noticia:
            # Aquí imprimimos la noticia obtenida
            print("📰 Noticia encontrada:")
            print(noticia)

        # Validación IA
        validacion = validar_noticia(noticia, noticias_guardadas)
        puntuacion = validacion.get("puntuacion", 0)
        noticia["validacion"] = validacion
        noticias_obtenidas.append(noticia)

        print(f"✅ Noticia validada con puntuación {puntuacion}: {validacion.get('comentario')}")

        # Agrego URL a guardadas para no repetir
        noticias_guardadas.append(noticia["url"])

        # Si encuentro una noticia muy buena, la devuelvo inmediatamente
        if puntuacion >= 7:
            return noticia

    # Si ninguna noticia supera la puntuación mínima, devuelvo la mejor
    if noticias_obtenidas:
        mejor_noticia = max(
            noticias_obtenidas,
            key=lambda n: (n["validacion"].get("puntuacion", 0), n.get("viralidad", 0))
        )
        print(f"⚠️ Ninguna noticia superó la puntuación mínima, se devuelve la mejor: {mejor_noticia['title']}")
        return mejor_noticia

    return None
  
# Obtengo noticias de fitness
def obtener_noticia_newsapi(noticias_guardadas):
    url_noticias = "https://newsapi.org/v2/everything"
  
    # Uso un buscador orientado a contenido general saludable y fitness
    params = {
        "q": "fitness OR ejercicio OR entrenamiento OR salud OR deporte OR gimnasio OR musculación OR bienestar OR estiramientos OR postura",
        "language": "es",
        "sortBy": "publishedAt",
        "apiKey": NEWS_API_KEY,
        "pageSize": 20
    }
    # Palabras que NO quiero en noticias para redes sociales
    EXCLUSIONES_TITULO = {
      # Política y deportes mayores
      "política", "gobierno", "elecciones", "partido",
      "fútbol", "baloncesto", "nba", "mundial", "boxeo",
      
      # Marcas deportivas
      "nike", "adidas", "reebok", "under armour", "puma",
      "decathlon", "asics", "gymshark", "myprotein",
      
      # Cadenas de gimnasios
      "mcfit", "basic fit", "vivagym", "altafit", "planet fitness",
      
      # Productos comerciales
      "creatina", "whey", "proteína", "suplemento", "suplementos",
      "batido", "cápsulas", "vitaminas", "oferta", "rebajas",
      
      # Influencers (para evitar noticias vacías)
      "influencer", "instagram", "tiktoker", "celebridad",
      
      # Otros
      "empresa", "marca", "publicidad", "patrocinado"
    }

    try:
        response = requests.get(url_noticias, params=params)
        response.raise_for_status()  # Lanzo un error si el código de estado no es 200
        noticias = response.json().get("articles", [])
        
        noticias_filtradas = [
            {"title": n["title"], "url": n["url"], "image": n.get("urlToImage")}
            for n in noticias
            if not any(excluido in n["title"].lower() for excluido in ["política", "fútbol", "baloncesto", "empresa", "corrupción", "gobierno", "partido", "elecciones", "nba", "boxeo", "mundial"])
        ]
        
        if noticias_filtradas:
            # Recorro la lista de noticias para encontrar la más adecuada y que no haya sido publicada
            for noticia in noticias_filtradas:
                if noticia["url"] not in noticias_guardadas:
                    return noticia
            print("❌ No hay noticias nuevas para publicar.")
    except requests.exceptions.RequestException as e:
        print(f"Error al obtener noticias: {e}")
    return None

# Obtener noticias desde GNews
def obtener_noticia_gnews(noticias_guardadas):
    url_noticias = "https://gnews.io/api/v4/search"
    params = {
        "q": "fitness OR ejercicio OR entrenamiento OR salud OR deporte OR gimnasio OR musculación OR bienestar OR estiramientos OR postura",
        "lang": "es",
        "sortby": "publishedAt",
        "max": 20,
        "apikey": GNEWS_API_KEY
    }

    try:
        response = requests.get(url_noticias, params=params)
        response.raise_for_status()
        noticias = response.json().get("articles", [])

        noticias_filtradas = [
            {"title": n["title"], "url": n["url"], "image": n.get("image")}
            for n in noticias
            if not any(excluido in n["title"].lower() for excluido in [
                "política", "fútbol", "baloncesto", "empresa", "corrupción",
                "gobierno", "partido", "elecciones", "nba", "boxeo", "mundial"
            ])
        ]

        for noticia in noticias_filtradas:
            if noticia["url"] not in noticias_guardadas:
                return noticia
        print("❌ No hay noticias nuevas para publicar en GNews.")
    except requests.exceptions.RequestException as e:
        print(f"Error GNews: {e}")
    return None

# Función para validar la noticia con IA
def validar_noticia(noticia, noticias_guardadas):
    urls_guardadas = [
        n["url"] if isinstance(n, dict) and "url" in n else n
        for n in noticias_guardadas
    ]

    prompt = f"""
    Actúa como evaluador de contenido para HabitoFit, empresa de fitness con app y comunidad en redes.
    Evalúa la siguiente noticia REAL sobre fitness, deporte, bienestar o salud:

    Título: {noticia['title']}
    URL: {noticia['url']}

    INSTRUCCIÓN ESPECIAL:
    - Si la noticia es alguna de estas noticias guardadas ({urls_guardadas}), 
      devuelve directamente puntuacion: 0 y viralidad: 0 sin hacer otra evaluación.
    
    PUNTUACIÓN:
    - Puntúa del 1 al 10 según interés para redes sociales de HabitoFit.
    - Criterios: relevancia, actualidad, valor para fitness/entrenamiento/bienestar, claridad y engagement potencial.
    - Explica brevemente la razón de la puntuación.

    SOLO DEVUELVE (en formato JSON literal):
    {{"puntuacion": 0-10, "viralidad": 0-10, "comentario": "explicación breve"}}
    """

    providers = [
      OpenRouterProvider(OPENROUTER_API_KEY),
      #OpenAIProvider(OPENAI_API_KEY),
    ]
  
    for provider in providers:
        try:
            resultado = provider.generate(prompt)
            # Intento convertir el resultado a dict
            return json.loads(resultado)
        except Exception as e:
            print(f"⚠️ Error validando noticia con {provider.__class__.__name__}: {e}")
            continue
    return {"puntuacion": 0, "comentario": "Error en validación"}
  
# ---------------------------AI PROVIDER---------------------------------
class AIProvider:
    def generate(self, prompt):
        raise NotImplementedError("Debes implementar este método en cada proveedor.")
        
class OpenAIProvider(AIProvider):
    def __init__(self, api_key):
        self.api_key = api_key

    def generate(self, prompt):
        url = "https://api.openai.com/v1/chat/completions"
        headers = {
            "Authorization": f"Bearer {self.api_key}",
            "Content-Type": "application/json",
        }

        data = {
            "model": "gpt-4o-mini",
            "messages": [{"role": "user", "content": prompt}],
            "max_tokens": 700
        }

        response = requests.post(url, headers=headers, json=data)
        response.raise_for_status()

        return response.json()["choices"][0]["message"]["content"].strip()

class OpenRouterProvider(AIProvider):
    def __init__(self, api_key, models=None, site_url=None, site_title=None):
        self.api_key = api_key
        self.site_url = site_url or "https://habitofit.com"
        self.site_title = site_title or "HabitoFit"

        # Lista de modelos por defecto (ordenados por calidad)
        self.models = models or [
            # OpenAI
            "openai/gpt-oss-120b:free",
            "openai/gpt-oss-20b:free",
        
            # Meta LLaMA
            "meta-llama/llama-4-maverick:free",
            "meta-llama/llama-4-scout:free",
            "meta-llama/llama-3.3-70b-instruct:free",
            "meta-llama/llama-3.2-3b-instruct:free",
        
            # Nous Research
            "nousresearch/hermes-3-llama-3.1-405b:free",
        
            # Google
            "google/gemini-2.0-flash-exp:free",
            "google/gemma-3-27b-it:free",
            "google/gemma-3-12b-it:free",
            "google/gemma-3n-e4b-it:free",
            "google/gemma-3n-e2b-it:free",
            "google/gemma-3-4b-it:free",
        
            # MistralAI
            "mistralai/mistral-small-3.1-24b-instruct:free",
            "cognitivecomputations/dolphin-mistral-24b-venice-edition:free",
            "mistralai/devstral-2512:free",
            "mistralai/mistral-7b-instruct:free",
        
            # DeepSeek / TNGTech
            "deepseek/deepseek-r1-distill-llama-70b:free",
            "tngtech/deepseek-r1t-chimera:free",
            "tngtech/tng-r1t-chimera:free",
        
            # NVIDIA
            "nvidia/nemotron-3-nano-30b-a3b:free",
            "nvidia/nemotron-nano-12b-v2-vl:free",
            "nvidia/nemotron-nano-9b-v2:free",
        
            # Alibaba
            "alibaba/tongyi-deepresearch-30b-a3b:free",
        
            # X-AI
            "x-ai/grok-4.1-fast:free",
        
            # Amazon
            "amazon/nova-2-lite-v1:free",
        
            # Z-AI
            "z-ai/glm-4.5-air:free",
        
            # MoonshotAI
            "moonshotai/kimi-k2:free",
        
            # Arcee-AI
            "arcee-ai/trinity-mini:free",
        
            # Qwen
            "qwen/qwen3-coder:free",
            "qwen/qwen3-4b:free"
        ]

        self.base_url = "https://openrouter.ai/api/v1/chat/completions"

    def generate(self, prompt):
        # Headers base para la API
        headers = {
            "Authorization": f"Bearer {self.api_key}",
            "Content-Type": "application/json",
        }
        if self.site_url:
            headers["HTTP-Referer"] = self.site_url
        if self.site_title:
            headers["X-Title"] = self.site_title

        # Probar modelos en cascada
        for model in self.models:
            try:
                payload = {
                    "model": model,
                    "messages": [{"role": "user", "content": prompt}],
                    "max_tokens": 700
                }

                response = requests.post(
                    url=self.base_url,
                    headers=headers,
                    data=json.dumps(payload),
                    timeout=30  # evitar colgados
                )

                response.raise_for_status()
                data = response.json()

                salida = data['choices'][0]['message']['content'].strip()
                if salida:
                    print(f"✅ OpenRouter generó usando modelo: {model}")
                    return salida

            except Exception as e:
                print(f"⚠️ Fallo en modelo {model}: {e}")
                continue

        raise RuntimeError("❌ Todos los modelos de OpenRouter fallaron.")
      
# ---------------------------Chat GPT----------------------------------
# Genero tweet para la noticia
def generar_post(noticia_publicacion):
    url_openai = "https://api.openai.com/v1/chat/completions"
    headers = {"Authorization": f"Bearer {OPENAI_API_KEY}", "Content-Type": "application/json"}

    titulo = noticia_publicacion.get('titulo', 'Sin título')
    medio = noticia_publicacion.get('medio', 'Medio desconocido')
    fecha = noticia_publicacion.get('fecha', 'Fecha no disponible')
    url = noticia_publicacion.get('url', '')
    resumen = noticia_publicacion.get('resumen', '')

    prompt = f"""
      Actúa como creador de contenido para HabitoFit con estilo periodístico, cercano y claro.
      
      Se te proporciona la siguiente noticia REAL y VERIFICADA:
      
      Título: {titulo}
      Fecha de publicación: {fecha}
      URL: {url}
      Resumen: {resumen}

      TAREA:
      Redacta un tweet informativo y cercano comentando la noticia. 
      - Empieza directamente con el hecho noticioso.
      - Aporta contexto breve y conecta la información con fitness, bienestar, hábitos saludables o salud mental.
      - Menciona HabitoFit solo si aporta valor real.
      
      FORMATO OBLIGATORIO:
      - Texto principal del tweet.
      - Última línea del tweet: 1 o 2 hashtags relacionados con fitness o bienestar.
      
      REGLAS:
      - No inventar datos, fechas ni URLs.
      - No mencionar noticias antiguas.
      - No inventar citas ni declaraciones.
      - No saludar ni incluir frases de apertura genéricas.
    """

    providers = [
        OpenRouterProvider(OPENROUTER_API_KEY),
        #OpenAIProvider(OPENAI_API_KEY),
    ]

    max_retries = 5

    for provider in providers:
        for attempt in range(1, max_retries + 1):
            try:
                texto = provider.generate(prompt)

                if texto and texto.strip():
                    return texto
                    
            except Exception as e:
                print(f"❌ Error con {provider.__class__.__name__}: {e}")
                break  # Pasa al siguiente proveedor

    print("❌ Todos los proveedores fallaron o generaron ideas repetidas")
    return None
  
def ajustar_texto(texto, limite=256):
    # Encuentro todas las etiquetas que empiezan con #
    etiquetas = re.findall(r'#\w+', texto)

    # Mientras el texto supere el límite y haya etiquetas, elimino la última etiqueta
    while len(texto) > limite and etiquetas:
        etiqueta_a_eliminar = etiquetas.pop()  # tomo la última etiqueta
        # Elimino la etiqueta del texto (con su posible espacio anterior)
        texto = re.sub(r'\s*' + re.escape(etiqueta_a_eliminar), '', texto, count=1)

    # Elimino cualquier etiqueta inválida como '#' sola o '#' con espacios
    texto = re.sub(r'\s*#\s*', '', texto)
  
    return texto
  
# ---------------------------Twitter----------------------------------
# ------------------------------------------------------
# Publicar tweet usando Tweepy (hasta 3 intentos)
# ------------------------------------------------------
def publicar_tweet(texto, imagen_publicacion, intentos=3, espera=5):
    """
    Publica un tweet con manejo de duplicados, errores, autenticación y reintentos.
    Devuelve True si se publica con éxito, False en caso contrario.
    """
    for intento in range(1, intentos + 1):
        # Variar el texto para evitar bloqueos por duplicado
        post = limpiar_texto(post)
        print(post)
      
        try:
            media_ids = None
            if imagen_publicacion:
                # Si es path, subimos la imagen a Twitter
                media_response = client.media_upload(filename=imagen_publicacion)
                media_ids = [media_response.media_id_string]
                print(f"📤 Imagen subida con media_id: {media_ids}")

            # Publicamos el tweet con o sin imagen
            response = client.create_tweet(text=post, media_ids=media_ids)
            tweet_id = response.data['id']
            print(f"✅ Tweet publicado correctamente: {tweet_id}")
            return True

        except tweepy.TweepyException as e:
            code_http = getattr(e.response, 'status_code', None) if hasattr(e, 'response') else None
            detalle = getattr(e.response, 'text', None) if hasattr(e, 'response') else None

            print(f"⚠️ Intento {intento}/{intentos} - Código: {code_http} - Error: {e}")
            if detalle:
                print(f"🧩 Detalle del error: {detalle}")

            # Manejo de errores comunes
            if code_http == 403:
                print("❌ Error 403: el tweet podría ser duplicado o el token no tiene permisos de escritura.")
                break
            elif code_http == 429:
                print("⏳ Rate limit alcanzado. Esperando antes de reintentar...")
                time.sleep(espera * 3)
                espera *= 2
            elif code_http in (500, 502, 503):
                print(f"⚠️ Error del servidor ({code_http}). Reintentando en {espera}s...")
                time.sleep(espera)
                espera *= 2
            else:
                print("❌ Error no recuperable, saliendo del bucle.")
                break

    print("❌ No se pudo publicar el tweet después de los intentos.")
    return False


def limpiar_texto(texto):
    """
    Elimina emojis y caracteres invisibles o raros del texto.
    Mantiene los hashtags (#) normales.
    """

    # Patrón de emojis que NO incluye el "#"
    emoji_pattern = re.compile(
        "("
        "[\U0001F600-\U0001F64F]"   # emoticonos
        "|[\U0001F300-\U0001F5FF]"  # pictogramas
        "|[\U0001F680-\U0001F6FF]"  # transporte
        "|[\U0001F1E0-\U0001F1FF]"  # banderas
        "|[\U00002700-\U000027BF]"  # símbolos varios
        "|[\U0001F900-\U0001F9FF]"  # suplementarios
        "|[\U00002600-\U000026FF]"  # misceláneos
        "|[\U00002B00-\U00002BFF]"  # símbolos misceláneos
        "|[\U0001FA70-\U0001FAFF]"  # símbolos extra
        "|[\U000025A0-\U000025FF]"  # geometría
        "|[\uFE0F]"                 # variante emoji (importante)
        ")"
    )

    # Eliminamos solo emojis, NO elimina '#'
    texto_sin_emojis = emoji_pattern.sub('', texto)

    # Eliminar caracteres invisibles
    invisible_pattern = re.compile(
        r'[\u200B-\u200F\u202A-\u202E\u2060-\u206F\uFEFF]'
    )
    texto_limpio = invisible_pattern.sub('', texto_sin_emojis)

    # Mantener solo caracteres imprimibles (no afecta '#')
    texto_limpio = ''.join(c for c in texto_limpio if c.isprintable())

    return texto_limpio

def acortador_url(url_larga):
    """
    Acorta una URL usando TinyURL y devuelve la URL corta.
    """
    api = "http://tinyurl.com/api-create.php"
    try:
        response = requests.get(api, params={"url": url_larga}, timeout=5)
        if response.status_code == 200:
            return response.text
        else:
            print(f"⚠️ No se pudo acortar la URL, status code: {response.status_code}")
            return url_larga
    except Exception as e:
        print(f"⚠️ Error al acortar la URL: {e}")
        return url_larga
# ---------------------------BOT DISCORD----------------------------------
async def traducir_a_ingles_lingva(texto):
    texto_encoded = urllib.parse.quote(texto)
    url = f"https://lingva.ml/api/v1/es/en/{texto_encoded}"

    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                      "AppleWebKit/537.36 (KHTML, like Gecko) "
                      "Chrome/116.0.0.0 Safari/537.36"
    }

    async with aiohttp.ClientSession(headers=headers) as session:
        try:
            async with session.get(url, timeout=10) as res:
                if res.status == 200:
                    data = await res.json()
                    return data.get("translation", texto)
                else:
                    print(f"⚠️ Error traduciendo Lingva: {res.status}")
                    return texto
        except Exception as e:
            print(f"⚠️ Excepción traduciendo con Lingva: {e}")
            return texto

async def traducir_a_ingles_mymemory(texto):
    try:
        texto_encoded = urllib.parse.quote(texto)
        url = f"https://api.mymemory.translated.net/get?q={texto_encoded}&langpair=es|en"

        async with aiohttp.ClientSession() as session:
            async with session.get(url, timeout=10) as res:
                if res.status == 200:
                    data = await res.json()
                    traduccion = data.get("responseData", {}).get("translatedText")
                    return traduccion or texto
                else:
                    print(f"⚠️ Error MyMemory: {res.status}")
                    return texto
    except Exception as e:
        print(f"⚠️ Excepción en MyMemory: {e}")
        return texto


async def post_to_discord(texto, imagen_publicacion):
    try:
        # Conexión a Discord
        await discord_client.login(DISCORD_TOKEN_BOT)

        # Servidor
        guild = await discord_client.fetch_guild(1334848244894662757)
        if guild is None:
            print("❌ El bot no está en el servidor adecuado.")
            return

        # Canales
        canal_es = await guild.fetch_channel(1335953397651734560)
        canal_en = await guild.fetch_channel(1431590526666276974)
        if canal_es is None or canal_en is None:
            print("❌ No se encontraron los canales.")
            return

        # Enviar al canal en español
        with open(imagen_publicacion, "rb") as f:
            embed_es = discord.Embed(title=ajustar_texto_embed(texto))
            await canal_es.send(embed=embed_es, file=discord.File(f, "imagen_publicacion.jpg"))

        # Traducir texto al inglés
        texto_en = await traducir_a_ingles_lingva(texto)
        if texto_en == texto:  # Lingva falló o devolvió lo mismo
            texto_en = await traducir_a_ingles_mymemory(texto)

        # Enviar al canal en inglés
        with open(imagen_publicacion, "rb") as f:
            embed_en = discord.Embed(title=ajustar_texto_embed(texto_en))
            await canal_en.send(embed=embed_en, file=discord.File(f, "imagen_publicacion.jpg"))

        print("✅ Post enviado correctamente a ambos canales.")

    except Exception as e:
        print(f"❌ Ocurrió un error: {e}")

    finally:
        await discord_client.close()

def ajustar_texto_embed(texto, limite=256):
    if len(texto) <= limite:
        return texto
    return texto[:limite - 3] + '...'

async def obtener_imagen_pixabay(query):
    """Obtiene una imagen relevante desde Pixabay si la imagen de la noticia no está disponible."""
    async with aiohttp.ClientSession() as session:
        params = {
            "key": PIXABAY_API_KEY,
            "q": query,
            "image_type": "photo",
            "per_page": 1
        }
        async with session.get(PIXABAY_URL, params=params) as respuesta:
            if respuesta.status == 200:
                datos = await respuesta.json()
                if datos["hits"]:
                    return datos["hits"][0]["webformatURL"]
    return None

# ---------------------------Make (Linkedin, Facebook e Instagram)----------------------------------
def publicar_con_make(texto, imagen_publicacion):
    try:
        print("📥 Abriendo imagen local...")
        try:
            with open(imagen_publicacion, "rb") as f:
                contenido_imagen = f.read()
            print("✅ Imagen cargada correctamente.")
        except Exception as e:
            print(f"❌ Error al abrir la imagen: {e}")
            return

        try:
            print("🛠 Procesando imagen...")
            processed_image = procesar_imagen(contenido_imagen)
            print("✅ Imagen procesada correctamente.")
        except Exception as e:
            print(f"❌ Error al procesar la imagen: {e}")
            return

        try:
            print("☁️ Subiendo imagen a File.io...")
            url_temporal = subir_a_uguu(processed_image.getvalue())
            print(f"✅ Imagen subida. URL temporal: {url_temporal}")
        except Exception as e:
            print(f"❌ Error al subir a File.io: {e}")
            return

        file_name = 'image.jpeg'

        files = {
            'image': (file_name, processed_image.getvalue(), 'image/jpeg')
        }

        data = {
            'image_url': url_temporal,
            'content': texto
        }

        print("📤 Enviando POST a Make...")
        response = requests.post(webhook_url, files=files, data=data)

        if response.status_code == 200:
            print("✅ Post enviado correctamente a Make.")
        else:
            print("❌ Error al enviar el post:", response.status_code, response.text)

    except Exception as e:
        print(f"❌ Error al publicar con Make: {e}")


async def generar_imagen(noticia_publicacion):
    output_path = "/kaggle/working/imagen_post.png"

    titulo = noticia_publicacion.get('titulo', 'Sin título')
    image = noticia_publicacion.get('image', '')
    #image = "https://s03.s3c.es/imag/_v0/1200x655/3/c/0/telefonica-distrito2.jpg"

    if not image:
        print("No hay URL de imagen")
    else:
        headers = {
            "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                          "AppleWebKit/537.36 (KHTML, like Gecko) "
                          "Chrome/120.0.0.0 Safari/537.36"
        }

    
    # Descargo la imagen
    response = requests.get(image, headers=headers)
    
    if response.status_code == 200:
        # Guardo temporalmente
        temp_path = "/kaggle/working/temp_image"
        with open(temp_path, "wb") as f:
            f.write(response.content)
        
        # Detecto tipo real
        ext = imghdr.what(temp_path) or 'jpg'
        image_noticia_path = f"/kaggle/working/image_noticia.{ext}"
    
        # Renombro con la extensión correcta
        os.rename(temp_path, image_noticia_path)
        print(f"Imagen descargada correctamente como {image_noticia_path}")
    else:
        print("Error al descargar la imagen")
        return None

    prompt = (
        "Usa la imagen proporcionada como referencia PRINCIPAL para generar la nueva imagen. "
        "La composición, los objetos y los elementos visibles deben mantenerse, "
        "pero si hay personas humanas, varia ligeramente el ángulo, iluminación y perspectiva "
        "para evitar problemas de copyright. La imagen generada debe ser fiel al contenido, "
        "pero completamente original en ejecución.\n\n"
        
        f"Tema central inspirado en la noticia: {titulo}.\n\n"
        
        "🎨 Reglas según el contenido de la imagen:\n"
        "- Si aparece una persona famosa: mantén su identidad reconocible, pero cambia ligeramente su perspectiva, ángulo o iluminación.\n"
        "- Si aparece cualquier otro objeto o elemento: genera una imagen profesional para redes sociales que sea diferente a la imagen adjuntada.\n\n"
        
        "🔤 Texto: Agrega un overlay de color negro profesional en la parte inferior con algo de contexto del post.\n\n"
        "Usa el color blanco para el texto. Si vas a poner detalles que sean de color azul.\n\n"

        "📐 Especificaciones técnicas: alta resolución, aspect ratio 4:5 para Instagram feed, estilo limpio y profesional.\n\n"
    )


    try:
        resultado = await descargar_imagen_capcut(prompt, output_path, image_noticia_path)
        if resultado:
            return output_path
    except Exception as e:
        print(f"⚠️ Error al generar miniatura: {e}")
        return None

async def descargar_imagen_capcut(prompt, output_path, image_noticia_path):
    try:
        async with async_playwright() as p:
            # Configuro el navegador (usando Chromium)
            browser = await p.chromium.launch(headless=True)  # Inicio el navegador en modo headless
            context = await browser.new_context()

            # Cargo cookies
            cookies_path = '/kaggle/input/cookies-habitofit-capcut-2/cookies-habitofit-capcut.txt'

            # Cargo las cookies desde el archivo
            with open(cookies_path, "r", encoding="utf-8") as file:
                cookies = json.load(file)

            # Filtro e importar cookies para capcut.com
            for cookie in cookies:
                if ".capcut.com" not in cookie.get("domain", ""):
                    continue  # Ignoro cookies de otros dominios

                cookie_dict = {
                    "name": cookie.get("name"),
                    "value": cookie.get("value"),
                    "domain": cookie.get("domain"),
                    "path": cookie.get("path", "/"),
                    "secure": cookie.get("secure", False),
                    "httpOnly": cookie.get("httpOnly", False),
                }

                # Añado fecha de expiración si existe
                if "expirationDate" in cookie:
                    cookie_dict["expires"] = int(cookie["expirationDate"])

                await context.add_cookies([cookie_dict])

            # Navego al sitio y declino las cookies (si se pone el banner)
            page = await context.new_page()
            
            url_destino = "https://www.capcut.com/ai-homepage?from_page=work_space&start_tab=video&enter_from=page_header&tab=all"
            print(f"🌐 Navegando a: {url_destino}")
            
            try:
                # Intento la carga completa con timeout extendido
                await page.goto(url_destino, wait_until="load", timeout=90000)
                print("✅ Página cargada completamente.")
            except Exception:
                print("⚠️ La carga completa tardó demasiado, intentando con carga parcial (DOM)...")
                try:
                    await page.goto(url_destino, wait_until="domcontentloaded", timeout=90000)
                    print("✅ Página cargada parcialmente (DOM listo).")
                except Exception as e:
                    await page.screenshot(path="/kaggle/working/error_goto.png")
                    raise Exception(f"❌ Error crítico al cargar la página de CapCut: {e}")


            print("URL actual tras cargar:", page.url)

            await asyncio.sleep(5) # 5 segundos de espera
        
            await aceptar_terminos_servicio(page)
            
            # 🧩 Cerrar posibles modales que bloquean la vista
            modales = page.locator("div.lv-modal-wrapper")
            if await modales.count() > 0:
                print("⚠️ Se detectó un modal bloqueando la vista. Intentando cerrarlo...")
                try:
                    # Busca un botón de cerrar común (X o texto 'Cerrar')
                    boton_cerrar = page.locator(
                        "button:has-text('×'), button:has-text('Cerrar'), svg[aria-label='close']"
                    ).first
                    if await boton_cerrar.is_visible():
                        await boton_cerrar.click(timeout=5000)
                        await page.wait_for_timeout(1000)
                        print("✅ Modal cerrado correctamente.")
                    else:
                        raise Exception("No se encontró botón de cierre visible.")
                except Exception:
                    # Si no hay botón visible, eliminamos el modal con JS
                    await page.evaluate("""() => {
                        document.querySelectorAll('.lv-modal-wrapper').forEach(el => el.remove());
                    }""")
                    print("✅ Modal eliminado forzosamente.")
            else:
                print("✅ No se detectaron modales bloqueando.")
            
            # 🧹 Eliminar posibles overlays o modales que bloqueen clics
            await page.evaluate("""
                document.querySelectorAll('.lv-modal-mask, .modal-mask, [aria-hidden="true"]').forEach(el => el.remove());
            """)
            print("🧹 Overlay eliminado (si existía).")
            
            # 🎯 Localizar el div que contiene el span con el texto 'Diseño de IA' (sin sufijos)
            tab_diseno = page.locator("div:has(span:text-is('Diseño de IA'))").first
            
            # Esperar a que sea visible
            await tab_diseno.wait_for(state="visible", timeout=60000)
            
            # 🔧 Asegurar que sea clickeable (sin sufijos)
            await page.evaluate("""
                const el = [...document.querySelectorAll('div')]
                    .find(div => div.innerText.trim() === 'Diseño de IA');
                if (el) el.scrollIntoView({behavior: 'instant', block: 'center'});
            """)
            print("🟡 'Diseño de IA' encontrada, intentando clic JS directo...")
            
            # 🖱️ Clic real mediante JavaScript (ignora animaciones y overlays)
            await page.evaluate("""
                const el = [...document.querySelectorAll('div')]
                    .find(div => div.innerText.trim() === 'Diseño de IA');
                if (el) el.click();
            """)
            
            print("✅ 'Diseño de IA' clickeada correctamente por JS directo.")


            # Busco el textarea
            print("🔍 Buscando textarea o campo editable de la idea...")
            
            # Espera hasta 90s a que aparezca
            campo_prompt = None
            
            for i in range(90):
                campo_prompt = await page.evaluate_handle("""
                    () => {
                        const buscar = (root) => {
                            const posibles = Array.from(root.querySelectorAll('textarea, div[contenteditable="true"], input[type="text"]'));
                            for (const el of posibles) {
                                const texto = (el.getAttribute('placeholder') || el.innerText || el.parentElement?.innerText || '').toLowerCase();
                                if (texto.includes('idea') || texto.includes('prompt') || texto.includes('describe') || texto.includes('concepto')) {
                                    el.scrollIntoView({behavior:'smooth', block:'center'});
                                    return el;
                                }
                            }
                            for (const el of root.querySelectorAll('*')) {
                                if (el.shadowRoot) {
                                    const found = buscar(el.shadowRoot);
                                    if (found) return found;
                                }
                            }
                            return null;
                        };
                        return buscar(document);
                    }
                """)
                if campo_prompt:
                    break
                await asyncio.sleep(1)
            
            if not campo_prompt:
                await page.screenshot(path="/kaggle/working/error_textarea.png")
                raise Exception("❌ No se encontró el campo editable tras 90 segundos (ni en DOM ni en shadow DOM).")
            
            print("🧠 Intentando localizar y enfocar el campo de prompt...")
        
            playwright_timeout = 20000  # 20s de margen razonable
        
            try:
                # ⚙️ Intentar eliminar overlays o modales que bloqueen el click
                await page.evaluate("""
                    document.querySelectorAll('.lv-modal-mask, .modal-mask, [aria-hidden="true"]').forEach(el => {
                        el.style.display = 'none';
                        el.remove();
                    });
                """)
                print("🧹 Overlay eliminado (si existía).")
        
                # ✅ Localizador sin sufijos dinámicos
                campo_prompt_locator = page.locator("textarea[placeholder*='Introduce tus ideas'], textarea, div[contenteditable='true']").first
        
                # Esperar a que esté visible y estable
                await campo_prompt_locator.wait_for(state="visible", timeout=playwright_timeout)
                await campo_prompt_locator.scroll_into_view_if_needed()
        
                # Hacer click y forzar enfoque
                await campo_prompt_locator.click(force=True, timeout=playwright_timeout)
                await campo_prompt_locator.focus()
                print("✅ Campo editable detectado, enfocado y listo para escribir.")
        
                # ✏️ Escribir el prompt
                await campo_prompt_locator.fill(prompt)
                print("✏️ Prompt de miniatura introducido correctamente ✅")
        
            except Exception as e:
                print(f"❌ Error al intentar escribir el prompt: {e}")
                await page.screenshot(path="ERROR_escribir_prompt.png", full_page=True)
                print("📸 Captura guardada como 'ERROR_escribir_prompt.png' para análisis.")

            
            async with page.expect_file_chooser() as fc_info:
                await page.click('div[role="button"]')
            
            file_chooser = await fc_info.value
            await file_chooser.set_files(image_noticia_path)
                
            print("🔍 Buscando botón 'Enviar'...")
            
            boton_enviar = page.locator(
                'div[class^="rightActions"] button[type="button"]:not([disabled])'
            ).first
            
            # Espera a que esté habilitado
            await boton_enviar.wait_for(state="visible", timeout=60000)
            
            # Click seguro
            await boton_enviar.click()
            

            
            # 🧩 Guardamos la lista inicial de páginas
            paginas_antes = context.pages
            url_inicial = page.url
            
            screenshot_path = "/kaggle/working/preview_envio.png"
            await page.screenshot(path=screenshot_path)
            print(f"📸 Captura guardada en: {screenshot_path}")
            
            # 🚀 Clic en el botón de enviar
            await boton_enviar.click(force=True)
            print("🖱️ Botón 'Enviar' presionado, esperando resultado...")
            
            new_page = page  # asumimos misma pestaña, salvo que detectemos otra
            url_inicial = page.url
            resultado_detectado = False
            
            for i in range(90):  # Esperar hasta 90 segundos
                await asyncio.sleep(1)
                paginas_despues = context.pages
                nuevas = [p for p in paginas_despues if p not in paginas_antes]
            
                # ✅ Si aparece una nueva pestaña
                if nuevas:
                    new_page = nuevas[0]
                    print("🌐 Nueva pestaña detectada correctamente ✅")
                    break
            
                # ✅ Si cambia la URL en la misma pestaña
                if page.url != url_inicial:
                    print("🔄 Misma pestaña, pero URL cambió correctamente ✅")
                    break
            
                # ✅ Si aparece contenido generado en la misma página
                if await page.locator("button:has-text('Descargar')").count() > 0 or \
                   await page.locator("div:has-text('Descargar')").count() > 0 or \
                   await page.locator("canvas").count() > 0 or \
                   await page.locator("img").count() > 0:
                    resultado_detectado = True
                    print("🎨 Resultado generado detectado en la misma pestaña ✅")
                    break
            
            # ⚠️ Si no se detecta nada tras 90 segundos, capturamos pantalla y lanzamos error
            if not resultado_detectado and (new_page == page and page.url == url_inicial):
                await page.screenshot(path="/kaggle/working/error_envio.png")
                raise Exception("❌ No se detectó nueva pestaña, cambio de URL ni resultado tras 90 segundos.")
            
            # ⏱️ Esperar que el contenido esté listo para interactuar
            await new_page.wait_for_load_state('domcontentloaded')
            print("📄 Página lista para continuar con la descarga ✅")

            await aceptar_terminos_servicio(new_page)
            
            # Captura de estado actual (diagnóstico)
            screenshot_path = f"/kaggle/working/estado_post_envio.png"
            await new_page.screenshot(path=screenshot_path)
            print(f"📸 Captura guardada en: {screenshot_path}")

            # 🧠 Esperar a que termine el estado "Thinking..."
            print("⏳ Esperando a que finalice el estado 'Thinking...'...")
            
            try:
                # Esperar a que el div con texto "IA de CapCut" aparezca,
                # lo que indica que el proceso de pensamiento terminó
                await new_page.locator("div:has-text('IA de CapCut')").wait_for(state="visible", timeout=60000)
            
                # Y asegurar que ya no existe el texto "Thinking..."
                await new_page.locator("div:has-text('Thinking...')").wait_for(state="detached", timeout=10000)
            
                print("✅ 'Thinking...' desapareció y 'IA de CapCut' está visible. Listo para continuar.")
            except Exception as e:
                print(f"⚠️ Timeout o error esperando el fin de 'Thinking...': {e}")
                await new_page.screenshot(path="ERROR_wait_thinking.png", full_page=True)
                print("📸 Captura guardada en 'ERROR_wait_thinking.png'.")

            try:
                # Espero hasta que la imagen FINAL esté disponible
                await new_page.locator(
                    'img[src^="blob:"]:not([class*="preloadImage"])'
                ).wait_for(state="visible", timeout=60000)
            
                print("✅ Imagen REAL detectada correctamente.")
            
            except Exception as e:
                print(f"⚠️ Error o timeout esperando la imagen final: {e}")
                await new_page.screenshot(path="ERROR_wait_image_final.png", full_page=True)
                print("📸 Captura guardada en 'ERROR_wait_image_final.png'.")

                # 🔁 Intento crear la miniatura metiendo el prompt manualmente
                try:
                    print("✏️ Intentando escribir el prompt nuevamente...")
                
                    # --- BUSCAR TEXTAREA DE FORMA ESTABLE ---
                    textarea = new_page.locator(
                        'textarea.lv-textarea:not(.lv-textarea-autosize-mirror)'
                    )
                    await textarea.wait_for(state="visible", timeout=10000)
                
                    # aseguramos que es interactuable
                    await textarea.scroll_into_view_if_needed()
                    await textarea.focus()
                
                    # limpiar (a veces queda basura)
                    try:
                        await textarea.fill("")
                    except:
                        pass
                
                    # escribo el prompt
                    await textarea.fill(prompt)
                    print("📝 Prompt escrito correctamente.")
                    
                    # Localizo el primer input de tipo file que acepte imágenes
                    file_input = page.locator('input[type="file"][accept*="image"]').first
                    
                    # Subo la imagen
                    await file_input.set_input_files(image_noticia_path)
                    print("📤 Imagen subida correctamente")
                    
                    # Espero al thumbnail
                    thumbnail = page.locator('input[type="file"] + div img')  # img dentro del div justo después del input
                    await thumbnail.first.wait_for(state="visible", timeout=60000)
                    print("🖼️ Imagen visible en la interfaz")
                
                    # --- BOTÓN DE ENVÍO SEGURO SIN CLASES DINÁMICAS ---
                    try:
                        fallback_btn = page.locator(
                            'button[type="button"]:not([disabled]) span:has-text("Send")'
                        ).first
                
                        await fallback_btn.wait_for(state="visible", timeout=10000)
                        await fallback_btn.click()
                        print("📤 Prompt enviado (botón Send estándar).")
                
                    except Exception as e:
                        print(f"❌ No se pudo enviar el prompt: {e}")
                        raise
                
                    # --- ESPERAR IMAGEN FINAL ---
                    await new_page.locator(
                        'img[src^="blob:"]:not([class*="preload"])'
                    ).wait_for(state="visible", timeout=1200000)
                
                    print("✅ Imagen REAL detectada correctamente.")

                    await asyncio.sleep(5) # 5 segundos de espera
                
                except Exception as inner_e:
                    print(f"❌ Error reenviando el prompt: {inner_e}")
                    await new_page.screenshot(path="ERROR_resend_prompt.png", full_page=True)
                    print("📸 Captura guardada en 'ERROR_resend_prompt.png'.")

            
            # 🔎 Buscar cualquier botón con texto "Descargar" o similar
            boton_descargar = new_page.locator("div[class^='exportArea'] button", has_text="Descargar").first
            await boton_descargar.wait_for(state="attached", timeout=90000)
            print("✅ Botón 'Descargar' detectado en el DOM.")
            
            # Intentar clic directamente con JS (sin requerir visibilidad)
            await new_page.evaluate("""
              const btn = document.querySelector("div[class^='exportArea'] button");
              if (btn) {
                  btn.scrollIntoView({behavior: 'smooth', block: 'center'});
                  btn.click();
              }
            """)
            print("🚀 Clic forzado en el botón 'Descargar' ejecutado.")
            
            print("🔍 Buscando el botón 'Descargar' dentro del diálogo de guardar...")

            # Captura de estado actual (diagnóstico)
            screenshot_path = f"/kaggle/working/BUSCOELBOTONDESCARGAR.png"
            await new_page.screenshot(path=screenshot_path)
            print(f"📸 Captura guardada en: {screenshot_path}")
        
            try:
                # Buscar un div cuya clase contenga 'upload-container' (sin depender del sufijo)
                dialogo_descargar = new_page.locator("div[class*='upload-container']")
                await dialogo_descargar.wait_for(state="visible", timeout=20000)
                print("✅ Diálogo de guardar detectado y visible.")
        
                # Buscar un botón con el texto 'Descargar' dentro de ese contenedor
                boton_descargar = dialogo_descargar.locator("button:has-text('Descargar')")
                await boton_descargar.wait_for(state="visible", timeout=20000)
                print("✅ Botón 'Descargar' localizado correctamente.")
        
                # Capturar la descarga y hacer clic
                async with new_page.expect_download() as descarga_info:
                    await boton_descargar.click(force=True)
        
                descarga = await descarga_info.value
                await descarga.save_as(output_path)
                print(f"📥 Miniatura descargada correctamente en: {output_path}")
        
            except Exception as e:
                print(f"❌ Error al intentar descargar: {e}")
                await new_page.screenshot(path="error_descargar.png", full_page=True)
                print("📸 Captura de error guardada como 'error_descargar.png'.")
            
            return output_path
            
    except Exception as main_error:
        print(f"❌ Error al ejecutar Playwright: {main_error}")

    finally:
        # Cierro el navegador
        if 'browser' in locals():
            await browser.close()

async def aceptar_terminos_servicio(page):
    print("🔍 Buscando diálogo de Términos del Servicio...")

    try:
        # Esperar hasta que aparezca el modal
        await page.wait_for_function(
            "() => document.querySelector('.lv-modal-content')",
            timeout=30000
        )
        print("✅ Detectado modal en el DOM.")

        # Clic en 'OK' aunque esté oculto
        clicked = await page.evaluate("""
            () => {
                const btn = Array.from(document.querySelectorAll('button, span'))
                    .find(e => e.textContent.trim() === 'OK');
                if (btn) {
                    btn.style.display = 'block';
                    btn.style.visibility = 'visible';
                    btn.click();
                    return true;
                }
                return false;
            }
        """)

        if clicked:
            print("✅ Botón 'OK' clicado correctamente.")
        else:
            print("⚠️ No se encontró botón 'OK'.")

        # Esperar a que el modal desaparezca o esté oculto
        print("⏳ Esperando a que el diálogo se cierre o se oculte...")
        await page.wait_for_function(
            """() => {
                const el = document.querySelector('.lv-modal-content');
                if (!el) return true;
                const style = window.getComputedStyle(el);
                return style.display === 'none' || style.visibility === 'hidden' || style.opacity === '0';
            }""",
            timeout=15000
        )
        print("🎉 Diálogo cerrado o visualmente oculto.")

    except Exception as e:
        print(f"⚠️ No desapareció el diálogo: {e}")
        print("🧹 Eliminando manualmente los elementos...")
        await page.evaluate("""
            () => {
                document.querySelectorAll('.lv-modal-content, .lv-modal-header, .lv-modal-title')
                    .forEach(e => e.remove());
            }
        """)
        print("🧩 Diálogo eliminado manualmente del DOM.")

    try:
        await page.evaluate("document.body.style.transition='none'")
        await page.screenshot(path="ok_dialog_final_state.png", full_page=True)
        print("📸 Captura final guardada correctamente.")
    except Exception as e:
        print(f"⚠️ No se pudo tomar la captura final: {e}")

def procesar_imagen(image_data):
    MAX_FILE_SIZE = 8 * 1024 * 1024  # 8 MiB
    MIN_WIDTH = 320
    MAX_WIDTH = 1440
    MIN_ASPECT_RATIO = 0.8   # 4:5
    MAX_ASPECT_RATIO = 1.91  # 1.91:1

    # Cargar la imagen
    image = Image.open(BytesIO(image_data))

    # Convertir a JPEG si no lo es
    if image.format != 'JPEG':
        image = image.convert('RGB')

    width, height = image.size
    aspect_ratio = width / height

    # Ajustar proporción si está fuera de rango
    if aspect_ratio < MIN_ASPECT_RATIO:
        new_height = int(round(width / MIN_ASPECT_RATIO))
        top = max(0, (height - new_height) // 2)
        bottom = top + new_height
        image = image.crop((0, top, width, bottom))
    elif aspect_ratio > MAX_ASPECT_RATIO:
        new_width = int(round(height * MAX_ASPECT_RATIO))
        left = max(0, (width - new_width) // 2)
        right = left + new_width
        image = image.crop((left, 0, right, height))

    # Redimensionar si el ancho está fuera del rango permitido
    width, height = image.size
    if width < MIN_WIDTH:
        scale = MIN_WIDTH / width
        new_width = MIN_WIDTH
        new_height = int(round(height * scale))
        image = image.resize((new_width, new_height), Image.LANCZOS)
    elif width > MAX_WIDTH:
        scale = MAX_WIDTH / width
        new_width = MAX_WIDTH
        new_height = int(round(height * scale))
        image = image.resize((new_width, new_height), Image.LANCZOS)

    # Forzar dimensiones enteras finales
    final_width, final_height = image.size
    final_width = int(round(final_width))
    final_height = int(round(final_height))
    image = image.resize((final_width, final_height), Image.LANCZOS)

    # Mostrar dimensiones y aspecto
    aspect_ratio = final_width / final_height
    def decimal_to_ratio(w, h):
        divisor = gcd(w, h)
        return f"{w // divisor}:{h // divisor}"
    print(f"Imagen final: {final_width}x{final_height} px")
    print(f"Aspect ratio: {aspect_ratio:.3f} ({decimal_to_ratio(final_width, final_height)})")

    # Guardar en JPEG con compresión adaptativa
    quality = 85
    buffer = BytesIO()
    while True:
        buffer.seek(0)
        buffer.truncate()
        image.save(buffer, format='JPEG', quality=quality)
        size = buffer.tell()
        if size <= MAX_FILE_SIZE or quality <= 20:
            break
        quality -= 5

    if size > MAX_FILE_SIZE:
        raise ValueError("No se pudo comprimir la imagen por debajo de 8MiB.")

    buffer.seek(0)
    return buffer
  
def subir_a_uguu(image_bytes):
    files = {
        'files[]': ('image.jpeg', BytesIO(image_bytes), 'image/jpeg')
    }
    response = requests.post('https://uguu.se/upload', files=files)

    print("📡 Respuesta uguu.se:", response.status_code)
    print("🔍 Contenido respuesta:", response.text)

    if response.status_code == 200:
        data = response.json()
        if 'files' in data and len(data['files']) > 0:
            # El link directo está en data['files'][0]['url']
            return data['files'][0]['url']
        else:
            raise Exception(f"Respuesta inesperada de uguu.se: {data}")
    else:
        raise Exception(f"Error al subir a uguu.se: {response.status_code} - {response.text}")


# ---------------------------Lógica principal----------------------------------
noticias_guardadas = obtener_noticias_guardadas()
print(noticias_guardadas)
noticia_publicacion = obtener_noticia(noticias_guardadas)
texto_publicacion = generar_post(noticia_publicacion)
imagen_publicacion = await generar_imagen(noticia_publicacion)

print(f"Texto de la publicación: {texto_publicacion}")

# Si es fallback, termino el notebook
if imagen_publicacion is None:
    print("❌ Se produjo un fallback. Finalizando el notebook.")
    sys.exit(1)

# Muestro la imagen (si existe)
display(imagen_publicacion)


# Verifico si la función devolvió None
if texto_publicacion is None:
    print("Hubo un error al generar el tweet. No se procederá con la publicación.")
else:
    isPublicado = publicar_tweet(texto_publicacion, imagen_publicacion)

    # Ejecuto la función y cerrar el bot cuando termine
    asyncio.run(post_to_discord(ajustar_texto(texto_publicacion), imagen_publicacion))
  
    #texto += f"\n\n📰 Lee la noticia completa aquí: {noticia['url']}"
    publicar_con_make(texto_publicacion, imagen_publicacion)

    isPublicado = True
  
    if isPublicado:
        print(f"✅ Post enviado correctamente a Twitter.")
        guardar_noticia_publicada(texto_publicacion)  # Guardo la URL solo si el tweet se publicó correctamente
        print(f"✅ Noticia guardada: {texto_publicacion}")
    else:
        print("El tweet no se publicó, no se guardará la noticia.")
```
